```python
def sum_digits(n):

    if n < 10:
        return n

    return n % 10 + sum_digits(n // 10)
```

```python
print(sum_digits(1234))
```

n % 10   → get last digit <br>
n // 10  → remove last digit<br>
n < 10   → stop recursion <br>
<br><br>
sum_digits(1234)<br>
= 4 + sum_digits(123)<br>
= 4 + 3 + sum_digits(12)<br>
= 4 + 3 + 2 + sum_digits(1)<br>
= 4 + 3 + 2 + 1<br>
= 10

```python
def binary_search(Array, Key, Lowest=0, Highest=None):
    """
    Recursively searches for key in a sorted array.

    Returns:
        Index of key if found
        -1 if key is not found
    """
    if Highest is None:
        Highest = len(Array) - 1

    # Base case: empty search range
    if Lowest > Highest:
        return -1

    Midle = (Lowest + Highest) // 2

    # Key found
    if Array[Midle] == Key:
        return Midle

    # Search left half
    if Key < Array[Midle]:
        return binary_search(Array, Key, Lowest, Midle - 1)

    # Search right half
    return binary_search(Array, Key, Midle + 1, Highest)
```

```python

Array = [10, 20, 30, 40, 50, 60, 70]
key = 70

# Test
result = binary_search(Array, key)

print(result)
```

```python
def power_set(items):
    result = []

    def backtrack(index, current):
        # Base case: considered every item
        if index == len(items):
            result.append(current.copy())
            return

        # Choice 1: skip the current item
        backtrack(index + 1, current)

        # Choice 2: include the current item
        current.append(items[index])
        backtrack(index + 1, current)

        # Undo the choice
        current.pop()

    backtrack(0, [])
    return result

```

```python
items = [1, 2]

print(power_set(items))
```

```python
def hanoi(n, source, spare, destination):
    # Base case
    if n == 1:
        print(f"Move disk 1 from {source} to {destination}")
        return

    # 1. Move n-1 disks from source to spare
    hanoi(n - 1, source, destination, spare)

    # 2. Move the largest disk to destination
    print(f"Move disk {n} from {source} to {destination}")

    # 3. Move n-1 disks from spare to destination
    hanoi(n - 1, spare, source, destination)
```

```python
hanoi(3, "A", "B", "C")
```
