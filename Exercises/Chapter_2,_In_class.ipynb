```python
##Challenge 1
# Time: O(n), Space: O(n)
def is_balanced(s: str) -> bool:
    stack = []

    pairs = {
        ')': '(',
        ']': '[',
        '}': '{'
    }

    for char in s:
        if char in '([{':
            stack.append(char)

        elif char in ')]}':
            if not stack or stack.pop() != pairs[char]:
                return False

    return len(stack) == 0


print(is_balanced("(a[b]{c})"))   # True
print(is_balanced("([)]"))        # False
print(is_balanced("(("))          # False
print(is_balanced(""))            # True
print(is_balanced(")"))           # False
```

```python
##Challenge 2
# Time: O(n), Space: O(n)
def two_sum(nums: list[int], target: int) -> tuple[int, int]:
    seen = {}  # value -> index
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return (seen[complement], i)
        seen[num] = i


print(two_sum([2, 7, 11, 15], 9))        # (0, 1)
print(two_sum([-1, -2, -3, -4], -6))     # (1, 3)
print(two_sum([0, 4, 3, 0], 0))          # (0, 3)
```

```python
##Challenge 3

class Node:
    def __init__(self, val, next=None):
        self.val = val
        self.next = next


# Time: O(n), Space: O(1)
def reverse_list(head: Node | None) -> Node | None:
    prev = None
    curr = head

    while curr is not None:
        next_node = curr.next
        curr.next = prev
        prev = curr
        curr = next_node

    return prev


def print_list(head):
    curr = head
    while curr is not None:
        print(curr.val, end=" ")
        curr = curr.next
    print()


n3 = Node(3)
n2 = Node(2, n3)
n1 = Node(1, n2)

print_list(reverse_list(n1))        # 3 2 1
print_list(reverse_list(Node(7)))   # 7
print(reverse_list(None))           # None
```

```python
#Challenge 4
# Time: enqueue O(1), dequeue amortized O(1), Space: O(n)
class QueueFromStacks:
    def __init__(self):
        self.in_stack = []
        self.out_stack = []

    def enqueue(self, x) -> None:
        self.in_stack.append(x)

    def dequeue(self):
        if not self.out_stack:
            while self.in_stack:
                self.out_stack.append(self.in_stack.pop())
        if not self.out_stack:
            raise IndexError("dequeue from empty queue")
        return self.out_stack.pop()


q = QueueFromStacks()
q.enqueue(1)
q.enqueue(2)
q.enqueue(3)
print(q.dequeue())   # 1
print(q.dequeue())   # 2
q.enqueue(4)
print(q.dequeue())   # 3
print(q.dequeue())   # 4

try:
    q.dequeue()
except IndexError as e:
    print("Error:", e)
```

```python
##Challenge 5
# Time: O(n), Space: O(1)
def has_cycle(head: Node | None) -> bool:
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next          # 1 step
        fast = fast.next.next     # 2 steps

        if slow is fast:
            return True

    return False


# 1 -> 2 -> 3 -> 4 -> 5 -> back to 2
a1, a2, a3, a4, a5 = Node(1), Node(2), Node(3), Node(4), Node(5)
a1.next = a2
a2.next = a3
a3.next = a4
a4.next = a5
a5.next = a2
print(has_cycle(a1))      # True

# 1 -> 2 -> 3 (no cycle)
b3 = Node(3)
b2 = Node(2, b3)
b1 = Node(1, b2)
print(has_cycle(b1))      # False

print(has_cycle(None))    # False

c = Node(1)
print(has_cycle(c))       # False (single node)
c.next = c
print(has_cycle(c))       # True (self-loop)
```

```python
##Challenge 6
# Time: O(1) for every operation, Space: O(n)
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []   # min_stack[i] = smallest value in stack[0..i]

    def push(self, x) -> None:
        self.stack.append(x)
        if not self.min_stack:
            self.min_stack.append(x)
        else:
            self.min_stack.append(min(x, self.min_stack[-1]))

    def pop(self) -> None:
        if not self.stack:
            raise IndexError("pop from empty stack")
        self.stack.pop()
        self.min_stack.pop()

    def top(self):
        if not self.stack:
            raise IndexError("top from empty stack")
        return self.stack[-1]

    def get_min(self):
        if not self.min_stack:
            raise IndexError("get_min from empty stack")
        return self.min_stack[-1]


s = MinStack()
s.push(3)
s.push(5)
s.push(2)
print(s.get_min())   # 2
s.pop()
print(s.get_min())   # 3
print(s.top())       # 5

s.pop()
s.pop()
try:
    s.pop()
except IndexError as e:
    print("Error:", e)
```

```python
##Challenge 7
# Time: append O(1) amortized, get/set O(1), Space: O(n)
class DynamicArray:
    def __init__(self):
        self.capacity = 2
        self.size = 0
        self.data = [None] * self.capacity    # fixed-size backing array

    def append(self, x) -> None:
        if self.size == self.capacity:
            self._grow()
        self.data[self.size] = x
        self.size += 1

    def _grow(self):
        new_capacity = self.capacity * 2      # double the capacity
        new_data = [None] * new_capacity
        for i in range(self.size):            # copy old items
            new_data[i] = self.data[i]
        self.data = new_data
        self.capacity = new_capacity

    def get(self, i):
        if i < 0 or i >= self.size:
            raise IndexError("index out of range")
        return self.data[i]

    def set(self, i, x) -> None:
        if i < 0 or i >= self.size:
            raise IndexError("index out of range")
        self.data[i] = x


arr = DynamicArray()
arr.append(10)
arr.append(20)
print(arr.capacity)   # 2
arr.append(30)        # full -> grows
print(arr.capacity)   # 4
print(arr.get(2))     # 30

arr.set(0, 99)
print(arr.get(0))     # 99

try:
    arr.get(3)
except IndexError as e:
    print("Error:", e)
```

```python
##Challenge 8

class TreeNode:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


# Time: O(n), Space: O(h)
def is_valid_bst(root: TreeNode | None) -> bool:
    # Assumption: BST values must be distinct.
    def check(node, low, high):
        if node is None:
            return True

        # Current value must be strictly between its bounds.
        if not (low < node.val < high):
            return False

        # Every node in the left subtree must be smaller.
        # Every node in the right subtree must be larger.
        return (
            check(node.left, low, node.val)
            and check(node.right, node.val, high)
        )

    return check(root, float("-inf"), float("inf"))


valid = TreeNode(5, TreeNode(1), TreeNode(8, TreeNode(6), TreeNode(9)))
invalid = TreeNode(5, TreeNode(1), TreeNode(8, TreeNode(4), TreeNode(9)))

print(is_valid_bst(valid))        # True
print(is_valid_bst(invalid))      # False
print(is_valid_bst(None))         # True
print(is_valid_bst(TreeNode(1)))  # True
```
